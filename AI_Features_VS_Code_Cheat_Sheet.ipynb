```python

Visual Studio Code provides AI-powered autonomous agents, inline suggestions, chat, and smart actions. Agents plan, implement, and verify changes across multiple files and run in parallel, locally, in the background, or in the cloud. Choose from multiple AI models, connect to external tools with MCP, and customize agents for your team's workflows.

> Tip: If you don't yet have a Copilot subscription, sign up for the Copilot Free plan to get a monthly allowance of inline suggestions and AI credits.
>
> **Important:** Starting April 20, 2026, new sign-ups for Copilot Pro, Copilot Pro+, Max, and Student plans are temporarily paused.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 1. Essential keyboard shortcuts

- `Ctrl+Alt+I` - Open the Chat view
- `Ctrl+I` - Enter voice chat prompt in Chat view
- `Ctrl+N` - Start a new chat session in Chat view
- `Ctrl+Shift+I` - Switch to using agents in Chat view
- `Ctrl+I` - Start inline chat in the editor or terminal
- `Ctrl+I` (hold) - Start inline voice chat
- `Tab` - Accept inline suggestion or navigate to the next edit suggestion
- `Escape` - Dismiss inline suggestion
</VSCode.Cell>
<VSCode.Cell language="markdown">## 2. Access AI in VS Code

- **Chat view** (`Ctrl+Alt+I`): keep an ongoing chat conversation in the Secondary Side Bar.
- **Inline chat** (`Ctrl+I`): ask questions from the editor or terminal while staying in the flow.
- **Quick Chat** (`Ctrl+Shift+Alt+L`): ask quick questions without leaving your current task.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 3. Chat experience in VS Code

- Use the **Chat view** for a full conversation and context history.
- **Session controls** let you start new chat sessions, fork sessions, and review history.
- Use the **model picker** to switch AI models for chat.
- **Context window indicator** shows token usage and breakdown by category.
- **Queue or steer** lets you send a follow-up message while a request is running.
- **Voice** lets you speak prompts and hear responses.
- Enable **KaTeX** or **Mermaid** rendering for math and diagrams.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 4. Add context to your chat prompt

- Click **Add Context** to attach workspace files, symbols, selections, or terminal output.
- Drag and drop files, folders, or problem items into the Chat view.
- Use `#<file|folder|symbol>` to reference a file, folder, or symbol name in chat.
- Use `#-mention` to add a specific tool or context type to your prompt.
- Attach context from source control commits, test failures, or selected code.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 5. Chat tools and slash commands

### Key built-in tools
- `#agent` - delegate tasks to other agents.
- `#agent/runSubagent` - run a task in an isolated subagent context.
- `#browser` - interact with browser pages in the integrated browser.
- `#edit` - modify files in the workspace.
- `#execute` - run shell commands, terminal commands, or notebook cells.
- `#githubRepo` - search GitHub repositories semantically.
- `#githubTextSearch` - text search a GitHub repository or organization.
- `#read` - read workspace files, notebook cells, terminal history, or problems.
- `#search` - search workspace files, text, changes, or symbol usages.
- `#todos` - manage a todo list for the task.
- `#vscode/*` - use VS Code-specific commands and extension APIs.
- `#web/fetch` - fetch web page content from the current browser environment.

### Useful slash commands
- `/explain` - explain code or concepts.
- `/fix` - fix code or resolve lint/compiler errors.
- `/tests` - generate tests for selected methods or the whole file.
- `/setupTests` - get help setting up a testing framework.
- `/clear` - start a new chat session.
- `/compact` - summarize and compact the conversation.
- `/fork` - fork the chat session into a new thread.
- `/debug` - show chat debug logs.
- `/troubleshoot` - analyze agent debug logs for issues.
- `/new` - scaffold a new project or file.
- `/newNotebook` - scaffold a new notebook.
- `/plan` - create a detailed implementation plan.
- `/search` - generate a search query for the Search view.
- `/startDebugging` - create and start a debugging configuration.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 6. Using agents in VS Code

- Agents can plan, implement, and verify changes across files.
- Choose tools available to the agent, including built-in tools, MCP servers, and extension-provided tools.
- Use permission levels like Default Approvals, Bypass Approvals, or Autopilot to control behavior.
- Connect to MCP servers to extend capabilities with external tools and workflows.
- Use third-party agents like **Claude Agent (Preview)** and **OpenAI Codex** for alternative AI coding experiences.
- Use a **Plan agent** before implementation to create a structured plan and hand it off.
- Add tools such as terminal access, file editing, and search to improve agent results.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 7. Customize your chat experience

- Create **custom instructions** to define how the AI should operate for your team.
- Use **reusable prompt files** for common tasks like code generation, review, or testing.
- Build **custom agents** with pre-configured tools, instructions, and workflows.
- Use the **Agent Customizations editor** to manage instructions, prompts, hooks, and agents.
- Store workspace-level customizations to share with your team.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 8. Editor AI features

- **Inline suggestions** appear as you type, and `Tab` accepts suggestions.
- **Inline chat** (`Ctrl+I`) lets you ask questions from the editor.
- Use code comments to provide prompts directly in your source code.
- Access **context menu actions** for explain, fix, generate tests, and review selected code.
- Use **code actions** (lightbulb) to fix lint and compiler errors.
- Use `F2` for AI-powered symbol renaming suggestions.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 9. Source control, issues, and code review

- Use AI to generate commit messages, PR titles, and descriptions.
- Add source control changes or commits as chat context.
- Get help with merge conflicts and PR review feedback.
- Use the `@github` participant to ask about issues, pull requests, and repository state.
- Use review tools to analyze uncommitted changes and apply suggested fixes.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 10. Generate tests and debug problems

- `/tests` generates tests for methods or selected code.
- `/setupTests` helps configure a testing framework.
- `/fixTestFailure` suggests fixes for failing tests.
- `/fix` helps resolve compiler, lint, and runtime issues.
- `/startDebugging` generates launch configurations and starts debugging.
- AI can assist with issue diagnosis, config setup, and debugging workflows.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 11. Scaffold new projects and notebooks

- `/new` scaffolds a new project or file from natural language.
- `/newNotebook` creates a new Jupyter notebook scaffold.
- Agents can generate project structure, files, and documentation in one workflow.
- Use project scaffolding prompts to preview changes before applying them.
</VSCode.Cell>
<VSCode.Cell language="markdown">## 12. Terminal, Python, and notebook support

- Use terminal inline chat to ask shell and command questions.
- Use the `@terminal` participant for integrated terminal help.
- Use inline chat in the Native Python REPL to generate and run Python commands.
- Use notebook chat and attached variables to work directly with kernel state.
- Use `Ctrl+Alt+I` in notebooks to open chat and modify notebook content with AI.
```
